## Data Clean

***Menu***

```python
menu_df = pd.read_csv("raw data/Menu.csv", encoding='utf-8')
```

```python
# Drop null columns
Menu_Empty_Cols_to_Drop = ['keywords', 'language', 'location_type'] # These 3 columns are 100% empty
menu_df = menu_df.drop(Menu_Empty_Cols_to_Drop, axis=1)

# Drop Redundant columns
Menu_duplicate_cols = ['name'] # menu.'name' column is 82% null, and mostly equal to the values in sponsor table
menu_df = menu_df.drop(Menu_duplicate_cols, axis=1)

# Apply common data clean for str columns
menu_df['sponsor'] = menu_df['sponsor'].apply(data_clean_helper.clean_str_col)
menu_df['location'] = menu_df['location'].apply(data_clean_helper.clean_str_col)
menu_df['notes'] = menu_df['notes'].apply(data_clean_helper.clean_notes) # Kept semicolon as speparator 

# Further clean string columns
sponsor_map = {
    'restaurant and/or location not given': '',
    'restaurant and/or location unknown': ''
}
menu_df['sponsor'] = menu_df['sponsor'].replace(sponsor_map)
menu_df['location'] = menu_df['location'].replace(sponsor_map)


# Covert date column to proper datetime
menu_df['date'] = pd.to_datetime(menu_df['date'], errors='coerce')
invalid_dates = menu_df[menu_df['date'].isna() & menu_df['date'].notna()]
print(f"Validation - There are {len(invalid_dates)} invalid date value found.")
```

***Menu Page***

```python
menuPage_df = pd.read_csv("raw data/MenuPage.csv", encoding='utf-8')
```

```python
# Ensure the column is numeric
menuPage_df['page_number'] = pd.to_numeric(menuPage_df['page_number'], errors='coerce')
# Count how many values are not integer-like
non_integer_count = menuPage_df['page_number'].dropna().apply(lambda x: x % 1 != 0).sum()
print(f"Validation - page_number column - There are {non_integer_count} values found as not integer (exclude NaN).")

menuPage_df['page_number'] = menuPage_df['page_number'].astype('Int64')
print(f"Clean - page_number column - Converted dtype to int.")
```

***Menu Item***

```python
menuItem_df = pd.read_csv("raw data/MenuItem.csv", encoding='utf-8')
```

```python
# Apply the check
menuItem_df['not_utc'] = menuItem_df['created_at'].apply(
    lambda x: isinstance(x, str) and 'utc' not in x.lower()
)
non_utc_rows = menuItem_df[menuItem_df['not_utc']]
print(f"Validation - There are {len(non_utc_rows)} rows from created_at column found as not UTC timezone")

menuItem_df['not_utc'] = menuItem_df['updated_at'].apply(
    lambda x: isinstance(x, str) and 'utc' not in x.lower()
)
non_utc_rows = menuItem_df[menuItem_df['not_utc']]
print(f"Validation - There are {len(non_utc_rows)} rows from updated_at column found as not UTC timezone")
menuItem_df = menuItem_df.drop(columns = ['not_utc'])

# Convert both timestamp column to date
menuItem_df['created_at'] = pd.to_datetime(menuItem_df['created_at'], utc=True).dt.date
menuItem_df['updated_at'] = pd.to_datetime(menuItem_df['updated_at'], utc=True).dt.date
print(f"\nClean - Coverted two timestamps column to date.")
```

```python
# Ensure the column is numeric
menuItem_df['dish_id'] = pd.to_numeric(menuItem_df['dish_id'], errors='coerce')
# Count how many values are not integer-like
non_integer_count = menuItem_df['dish_id'].dropna().apply(lambda x: x % 1 != 0).sum()
print(f"Validation - dish_id column - There are {non_integer_count} values found as not integer(exclude NaN).")

menuItem_df['dish_id'] = menuItem_df['dish_id'].astype('Int64')
print(f"Clean - page_number column - Converted dtype to int.")
```

### Validation: Merge datasets by IDs

```python
# Both Dish and menuItem table contains dish id
dish_df = dish_df.rename(columns = {'id': 'dish_id'})
menuItem_df = menuItem_df.rename(columns = {'id': 'menuItem_id'})

# - First check the coverage
unique_dishid_in_dish = set(dish_df['dish_id']) - set(menuItem_df['dish_id'])

print(f"There are {len(unique_dishid_in_dish)} unique dishes don't have a match in menuItem table.")
print(f"- which is {(round(len(unique_dishid_in_dish)/len(dish_df) * 100))}% of dishes in dish table.")
```

```python
# Both menuItem and menuPage table contains menu_page_id
menuPage_df = menuPage_df.rename(columns = {'id': 'menu_page_id'})

# - First check the coverage
unique_menuPage_in_menuItem = set(menuItem_df['menu_page_id'].unique()) - set(menuPage_df['menu_page_id'])

print(f"There are {len(unique_menuPage_in_menuItem)} unique menu items don't have a match in menuPage table.")
print(f"- which is {(round(len(unique_menuPage_in_menuItem)/len(menuItem_df) * 100))}% of menuItem in menu Item table.")
```

```python
# Both menuPage and menu table contains menu_id
menu_df = menu_df.rename(columns = {'id': 'menu_id'})

# - First check the coverage
unique_menu_in_menuPage = set(menuPage_df['menu_id']) - set(menu_df['menu_id'])

print(f"There are {len(unique_menu_in_menuPage)} unique menu items don't have a match in menuPage table.")
print(f"- which is {(round(len(unique_menu_in_menuPage)/len(menuPage_df) * 100))}% of menu in menu Item table.")
```
